## String Methods


#### lower()

Convert an uppercase string into lowercase.


```python
text = 'I LOVE PyThOn'
text.lower()
```

#### upper()

Convert a lowercase string into uppercase.


```python
text = 'i lover python'
text.upper()
```

#### title()

`title()` capitalizes the first letter of each word.


```python
text = 'i lOvE PyThOn'
text.title()
```

#### capitalize()

`capitalize()` makes the first character uppercase.


```python
text = 'i LOvE PyThOn'
text.capitalize()
```

#### swapcase()

`swapcase()` changes uppercase letters to lowercase and lowercase letters to uppercase.


```python
text = 'i LOvE PyThOn'
text.swapcase()
```

#### casefold()

`casefold()` normalizes text for case-insensitive comparison.


```python
text = 'i LOvE PyThOn'
text.casefold()
```

#### strip()

`strip()` removes leading and trailing whitespace or selected characters.


```python
text = '         i LOvE PyThOn       '
text.strip()
```

#### lstrip()

`lstrip()` removes leading whitespace or selected characters.


```python
text = '         i LOvE PyThOn       '
text.lstrip()
```

#### rstrip()

`rstrip()` removes trailing whitespace or selected characters.


```python
text = '         i LOvE PyThOn       '
text.rstrip()
```

#### find()

`find()` returns the first position of a substring, or `-1` when absent.


```python
text = 'i LOvE Python  '
text.find('Python')
```

```python
text = "I live in Ranchi Jharkhand"
text.find('ranchi')
```

#### index()

`index()` returns the position of the first matching value.


```python
text = "I live in Ranchi Jharkhand"
text.index('in')
```

#### count()

`count()` returns how many times a value appears.


```python
text = "I live in Ranchi Jharkhand"
text.count('i')
```

#### replace()

`replace()` returns a new string with selected text replaced.


```python
text = 'I love you'
text.replace('you', 'Python')
```

#### startswith()

`startswith()` checks whether a string begins with a given prefix.


```python
text = 'Python Programming'
text.startswith('P')
```

#### endswith()

`endswith()` checks whether a string ends with a given suffix.


```python
text = 'Python Programming'
text.startswith('g')
```

#### isalpha()

`isalpha()` checks whether every character is alphabetic.


```python
text = 'Python Programming'
text.isalpha()
```

#### isdigit()

`isdigit()` checks whether every character is a digit.


```python
text = 'Python Programming'
text.isdigit()
```

#### isalnum()

`isalnum()` checks whether every character is a letter or digit.


```python
text = "anshuman1234"
text.isalnum()
```

#### isspace()

`isspace()` checks whether every character is whitespace.


```python
text = " "
text.isspace()
```
