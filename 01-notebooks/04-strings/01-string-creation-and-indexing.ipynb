## String Creation And Indexing


#### Creating Single-Quoted Strings

Create and display a string using single quotes.


```python
text = 'This is a text in single quotes'
```

#### Creating Double-Quoted Strings

Create and display a string using double quotes.


```python
text = "This is a text in single quotes"
```

#### Creating Triple-Quoted Strings

Use triple quotes for multi-line string content.


```python
text = '''This is a text in single quotes'''
```

#### Positive String Indexing

Use suitable values to complete positive string indexing and display the result.


```python
name = 'Anshuman'
print(name[0])
print(name[2])
```

#### Negative String Indexing

Use suitable values to complete negative string indexing and display the result.


```python
name = 'Anshuman'
print(name[-1])
print(name[-4])
```
