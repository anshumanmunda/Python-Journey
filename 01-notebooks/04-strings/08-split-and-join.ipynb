## Split And Join


#### Splitting by Whitespace


```python
text = 'This is a paragraph sample text'

text.split()
```

#### Splitting with a Custom Separator


```python
date = '02-06-2026'

date.split('-')
```

#### Joining a List of Strings


```python
lst_date = ['02', '06', '2026']
'/'.join(lst_date)
```

#### Joining with a Custom Separator


```python
text = ['This', 'is', 'a', 'paragraph', 'sample', 'text']

' '.join(text)
```
