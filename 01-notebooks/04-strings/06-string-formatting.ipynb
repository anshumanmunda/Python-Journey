## String Formatting


#### Positional Formatting with format()


```python
name = 'Anshuman'
age = 22

print('My name is {} and I am {} years old'.format(name, age))
```

#### Indexed Formatting with format()


```python
city = 'Ranchi'
state = 'Jharkhand' 

print('I live in {0} state {1}'.format(city, state))
```

#### Named Formatting with format()


```python
name = 'Anshuman'
age = 22

print('My name is {name} and I am {age} years old'.format(age=age, name=name))
```
