## String Slicing


#### Basic String Slicing

Extract a selected part of a string.


```python
text = "I live in Ranchi Jharkhand"
text[10:16]
```

#### Slicing with Step

Use suitable values to complete slicing with step and display the result.


```python
text = "I live in Ranchi Jharkhand"
text[17: : 2]
```

#### Reverse String Slicing

Display string slicing in reverse order.


```python
text = "I live in Ranchi Jharkhand"
text[::-1]
```

#### Negative Index Slicing

Use suitable values to complete negative index slicing and display the result.


```python
text = "I live in Ranchi Jharkhand"
text[-9: : 1]
```
