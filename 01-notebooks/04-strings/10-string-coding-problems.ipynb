## String Coding Problems


#### Palindrome Check

Check whether `madam` reads the same forward and backward.


```python
text = 'madam'

if text == text[::-1]:
  print('Palindrome')
else:
  print('Not Palindrome')
```

#### Character Frequency

Count how many times each character appears in a string.


```python
text = 'Anshuman Munda'
seen = []

for ch in text:
  if ch not in seen and ch != ' ':
    print(f'{ch} -> {text.count(ch)}')
    seen.append(ch)
```

#### Vowel and Consonant Count


```python
text = 'Data Science'
seen = []
vowels = consonant = 0 

for ch in text:
  if ch not in seen and ch != ' ':
    if ch.lower()  in "aeiou":
      vowels += 1
    else:
      consonant += 1
  seen.append(ch)

print(f'Vowels = {vowels}\nConsonants = {consonant}')      
```

#### Word Count


```python
sen = "Python was invented by Guido van Rossum in the Netherlands. He began working on it in late 1989 and first released it on February 20, 1991"

words = sen.split(' ')
word_count = len(words)
print(f'{word_count} words & {len(sen)} characters')
```

#### Reverse Each Word

Display each word in reverse order.


```python
sen = "Python was invented by Guido van Rossum"
words_lst = sen.split()
reverse_word_lst = []

for word in words_lst:
  reverse_word_lst.append(word[::-1])

reverse_sen = ' '.join(reverse_word_lst)
reverse_sen
```
