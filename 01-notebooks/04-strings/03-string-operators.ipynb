## String Operators


#### String Concatenation

Join two strings to form `Python Programming`.


```python
str1 = 'Python'
str2 = ' Programming'

str1 + str2
```

#### String Repetition

Repeat a short string three times.


```python
text = 'hehe ' * 3
text
```

#### String Membership

`"Py" in "Python"` → `True`


```python
text = 'Python'

print('Py' in text)
```

#### String Comparison

Compare two string values and display the result.


```python
text1 = 'A'
text2 = 'a'

print(text1 > text2)
```
