## F Strings


#### Basic f-String


```python
x = 69
print(f'value = {x}')
```

#### Expressions Inside f-Strings


```python
n = 90
print(f'{n + n // 2}')
```

#### Using Variables in f-Strings

Apply Variables in f-Strings with a small representative example.


```python
marks = 90
subject  = "Maths"

print(f'I scored {marks} marks in {subject}')
```
