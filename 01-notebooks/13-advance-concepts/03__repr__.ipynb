`__repr__()` It provides a detailed developer output for debugging

```python
class Counter:
  def __init__(self, value) -> None:
    self.count = value

  def count_up(self):
    self.count += 1  

  def count_down(self):
    self.count -= 1 

  def __add__(self, other):
    return self.count + other.count  

  def __str__(self):
    return f'Count is equal to {self.count} '

  def __repr__(self):
    return f'Counter(count = {self.count})'


counter1 = Counter(10)   
counter1.count_up()
counter2 = Counter(20)  

print( counter1 + counter2 )
print(counter1)
print(repr(counter1))
```
