When you use an operator, Python looks for a corresponding
magic method on the object.

- Addition (+) maps to __add__
- Length (len()) maps to __len__

```python
result = 'Hello' + 'World'
result
```

```python
res = 'Hello'.__add__('World')
res
```

```python
data = len('Anshuman') 
data
```

```python
data = 'Anshuman'.__len__()
data
```

**Operation on custom calss**

```python
class Counter:
  def __init__(self) -> None:
    self.count = 1

  def count_up(self):
    self.count += 1  

  def count_down(self):
    self.count -= 1 



counter1 = Counter()   
counter2 = Counter()  

print( counter1 + counter2 )
```

Defining Custom `__add__` method.

```python
class Counter:
  def __init__(self, value) -> None:
    self.count = value

  def count_up(self):
    self.count += 1  

  def count_down(self):
    self.count -= 1 

  def __add__(self, other):
    return self.count + other.count  

  def __str__(self):
    return f'Count is equal to {self.count} '

counter1 = Counter(10)   
counter1.count_up()
counter2 = Counter(20)  

print( counter1 + counter2 )
print(counter1)
```
