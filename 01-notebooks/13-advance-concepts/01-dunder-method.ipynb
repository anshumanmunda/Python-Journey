`Dunder Methods` Dunder means double underscores these are reserved, special method in python, They are also known as Magic method Because they are often Call implicitly by Python behind he scenes rather than a called by a programmer.

Magic method maps the operation or the operator to the internal class behaviour

- One of the most common magic method is `__init__` method. It is used to initialise the instant member variables

```python
class rectangle:
  def __init__(self, width: int, height: int) -> None:
    self.width = width
    self.height = height

  def area(self) -> int:
    return self.width * self.height

rec = rectangle(10, 20)
result = rec.area()
print(f'Area of rectangle: {result}')
```
