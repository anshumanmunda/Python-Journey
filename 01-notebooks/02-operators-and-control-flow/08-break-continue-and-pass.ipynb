## Break Continue And Pass


#### break Statement

Stop a loop when the value reaches `5`.


```python
for i in range(10):
  if i == 5:
    break
  print(i, end=' ')
```

#### continue Statement

Skip the value `5` and display the remaining values.


```python
for i in range(10):
  if i == 5:
    continue
  print(i, end=' ')
```

#### pass Statement

Create a valid empty block for code that will be completed later.


```python
if True:
  pass
```

#### break in a Nested Loop

Use suitable values to complete break in a nested loop and display the result.


```python
for i in range(3):
  for j in range(2):
    if j == 1:
      break
    print('Hello')

  print()    
```

#### continue with Conditions

Use suitable values to complete continue with conditions and display the result.


```python
for i in range(20):
  if i % 2 == 0:
    continue
  print(i, end=' ')
```
