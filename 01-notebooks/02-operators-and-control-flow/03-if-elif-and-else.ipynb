## If Elif And Else


#### Simple if Statement

Check whether a number is positive.

Display a message only when the condition is true.


```python
num = 20

if num > 0:
  print('Positive') 
```

#### if-else Statement

Check whether a number is even or odd.

Display the matching result.


```python
num = 23

if num % 2 == 0:
  print('Even')
else:
  print('Odd')
```

#### if-elif-else Statement

Classify marks into grade ranges.

Display the appropriate grade.


```python
marks = 90

if marks >= 90:
  print('Grade A+')

elif marks >= 80:  
  print('Grade A')

elif marks >= 70:  
  print('Grade B')

elif marks >= 60:  
  print('Grade C')

elif marks >= 50:  
  print('Grade D')

elif marks >= 40:  
  print('Grade E')
  
else:  
  print('Fail')
```
