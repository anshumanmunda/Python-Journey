## Range Function


#### range() with Stop Value

Use suitable values to complete range() with stop value and display the result.


```python
for i in range(5):
  print(i)
```

#### range() with Start and Stop


```python
for i in range(2,11):
  print(i)
```

#### range() with Start, Stop, and Step



```python
for i in range(2, 21, 2):
  print(i, end=' ')
```

#### Reverse range()

Display range() in reverse order.


```python
for i in range(20, 2, -2):
  print(i, end=' ')
```
