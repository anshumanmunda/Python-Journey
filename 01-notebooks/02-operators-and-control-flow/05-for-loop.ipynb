## For Loop


#### Iterating over a String

Display each value in a string one by one.


```python
text = 'I love Python'

for chr in text:
  print(chr)
```

#### Iterating over a List

Display each value in a list one by one.


```python
lst = [10, 20, 30, 40, 50, 60]

for i in lst:
  print(i)
```

#### Iterating over a Tuple

Display each value in a tuple one by one.


```python
t = (10, 20, 30, 40, 50, 60)

for i in t:
  print(i)
```

#### Iterating over a Dictionary

Display each value in a dictionary one by one.


```python
d = {
  'name': 'Anshuman',
  'age': 22,
  'city': 'Ranchi'
}

for key in d:
  print(d[key])
```

#### Nested for Loop

Create one for loop inside another and display the result.


```python
for i in range(4):
  for j in range(4):
    if i<j:
      print('*', end=' ')
    else:
      print()
        
```
