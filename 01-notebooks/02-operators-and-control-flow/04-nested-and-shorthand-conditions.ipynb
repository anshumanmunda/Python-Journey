## Nested And Shorthand Conditions


#### Nested if Statement

Create one if statement inside another and display the result.


```python
age = 22
has_id = True

if age >= 18:
  if has_id:
    print('You can vote')  
```

#### Nested if-else Statement

Create one if-else statement inside another and display the result.


```python
age = 22
has_id = False

if age >= 18:
  if has_id:
    print('You can vote') 
  else:  
    print('You can not vote') 
```

#### Single-Line if

Use suitable values to complete single-line if and display the result.


```python
num = 54

result = 'Odd' if num & 1 else 'Even'
print(result)
```
