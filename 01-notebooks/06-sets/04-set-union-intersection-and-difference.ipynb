## Set Union Intersection And Difference


#### union()

Find all unique values present in either of two sets.


```python
s1 = {1,2,3,4,5,5,6, 7, 8, 9, 10}
s2 = {1, 3, 0 , 5, 6}

# s1 | s2
s1.union(s2)
```

#### intersection()

Find the values common to two sets.


```python
s1 = {1,2,3,4,5,5,6, 7, 8, 9, 10}
s2 = {1, 3, 0 , 5, 6}

s1.intersection(s2)
```

#### difference()

Find the values present in the first set but not in the second.


```python
s1 = {1,2,3,4,5,5,6, 7, 8, 9, 10}
s2 = {1, 3, 0 , 5, 6}

s1.difference(s2)
```

#### Using | for Union

Apply | for Union with a small representative example.


```python
s1 = {1,2,3,4,5,5,6, 7, 8, 9, 10}
s2 = {1, 3, 0 , 5, 6}

s1 | s2
```

#### Using & for Intersection

Apply & for Intersection with a small representative example.


```python
s1 = {1,2,3,4,5,5,6, 7, 8, 9, 10}
s2 = {1, 3, 0 , 5, 6}

s1 & s2
```

#### Using - for Difference

Apply - for Difference with a small representative example.


```python
s1 = {1,2,3,4,5,5,6, 7, 8, 9, 10}
s2 = {1, 3, 0 , 5, 6}

s1 - s2
```

#### Symmetric Difference

```python
s1 = {1,2,3,4,5,5,6, 7, 8, 9, 10}
s2 = {1, 3, 0 , 5, 6}

s1 ^ s2
```

```python
s1 = {1,2,3,4,5,5,6, 7, 8, 9, 10}
s2 = {1, 3, 0 , 5, 6}

s1.symmetric_difference(s2)
```
