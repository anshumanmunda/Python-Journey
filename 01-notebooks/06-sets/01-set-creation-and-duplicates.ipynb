## Set Creation And Duplicates


#### Creating an Empty Set

Create an empty set and display its data type.


```python
s = {}

print(s, type(s))
```

```python
s = set()

print(s, type(s))
```

#### Creating a Set with Values

Create set with values with suitable sample values and display the result.


```python
s = {10, 50, 60, 'Hello'}

print(s)
```

#### Automatic Duplicate Removal

Create a set from values containing duplicates.

Display the unique values.


```python
s = {10, 50, 60, 'Hello', 'hello', 60, 50, 10, 20}

print(s)
```

#### Creating a Set from a List

Create set from a list with suitable sample values and display the result.


```python
l = [10, 50, 60, 60, 50, 10,]
st = set(l)
print(st)
```
