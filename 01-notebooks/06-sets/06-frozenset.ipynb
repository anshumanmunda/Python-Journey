## Frozenset


#### Creating a frozenset

Create frozenset with suitable sample values and display the result.


```python
# Normal Set
s = {0, 1, 2, 3, 4, 5, 5, 6, 7, 8, 9, 10}
s.add('Hello')

s
```

```python
# Frozenset

st = frozenset({1,23,4,33,4,3})
print(st)
```

#### Membership in a frozenset


```python
88 in st
```

#### Set Operations with frozenset


```python
st
st.pop() 
st.add(56) 
```
