## Set Relational Methods


#### issubset()

`issubset()` checks whether every element of one set is present in another.


```python
s1 = {0, 1, 2, 3, 4, 5, 5, 6, 7, 8, 9, 10}
s2 = {1, 3, 0 , 5, 6}

s2.issubset(s1)
```

#### issuperset()

`issuperset()` checks whether one set contains every element of another.


```python
s1 = {0, 1, 2, 3, 4, 5, 5, 6, 7, 8, 9, 10}
s2 = {1, 3, 0 , 5, 6}

s1.issuperset(s2)
```

#### isdisjoint()

`isdisjoint()` checks whether two sets have no common elements.


```python
s1 = {0, 1, 2, 3, 4, 5, 5, 6, 7, 8, 9, 10}
s2 = {1, 3, 0 , 5, 6}

s1.isdisjoint(s2)
```
