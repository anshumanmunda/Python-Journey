## Set Input And Conversion


#### Taking Set Input


```python
st = set(input('Enter elements of set: ').split(' '))
print(st, type(st))
```

#### List to Set Conversion

Convert a suitable value into list to set form and display the result.


```python
l = [10, 20, 50, 60, 90, 40, 40, 20, 60]
st = set(l)
st
```

#### Tuple to Set Conversion

Convert a suitable value into tuple to set form and display the result.


```python
t = (10, 20, 40, 50, 60, 90)
st = set(t)
st
```

#### Set to List Conversion

Convert a suitable value into set to list form and display the result.


```python
st = {10, 20, 40, 50, 60, 90}
l = list(st)
print(l, type(l))
```
