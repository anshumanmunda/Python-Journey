## Set Add Update And Remove


#### add()


```python
st = {10, 20, 60, 30, 10, 3, 0}
st.add(19)
print(st)
```

#### update()


```python
st = {10, 20, 60, 30, 10, 3, 0}
st.update(['Hello', 'hii', 'yes'])
print(st)
```

#### remove()

`remove()` deletes the first matching value from a list.


```python
st = {10, 20, 60, 30, 10, 3, 0}
st.remove(0)
```

#### discard()


```python
st = {10, 20, 60, 30, 10, 3, 0}
st.discard(69)
```

#### pop()

`pop()` removes and returns an element.


```python
st = {10, 20, 60, 30, 10, 3, 0}
st.pop()
```

#### clear()

`clear()` removes all items from a mutable collection.


```python
st = {10, 20, 60, 30, 10, 3, 0}
st.clear()
```
