## Type Conversion


#### String to Integer Conversion

Example: change the text value 25 into the number 25.


```python
int('25')
```

#### String to Float Conversion

Example: change the text value 3.14 into a decimal number.


```python
float(3.14)
```

#### Integer to String Conversion

Example: change the number 25 into text.


```python
num = 56
str(num)
```

#### Float to Integer Conversion

Example: 7.9 → 7


```python
f = 7.9 
int(f)
```

#### Integer to Float Conversion

Example: 7 → 7.0


```python
num = 69
float(num)
```

#### Value to Boolean Conversion

Example: zero becomes false, while a non-zero value becomes true.


```python
value = 10
bool(value)
```

#### Boolean to Integer Conversion

Example: true becomes 1 and false becomes 0.


```python
b = True
int(b)
```

```python
b2 = False
int(b2)
```
