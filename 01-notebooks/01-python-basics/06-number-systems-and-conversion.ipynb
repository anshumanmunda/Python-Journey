## Number Systems And Conversion


#### Decimal to Binary Conversion

Example: 7 → 111


```python
d = 7
print(f'Binary = {bin(d)}')
```

#### Decimal to Octal Conversion

Example: 8 → 10


```python
o = 8
print(f'Octal = {oct(o)}')
```

#### Decimal to Hexadecimal Conversion

Example: 255 → ff


```python
h = 255
print(f'Hexadecimal = {hex(h)}')
```

#### Binary to Decimal Conversion

Example: 111 → 7


```python
b = 0b111
print(f'Decimal = {b}')
```

#### Octal to Decimal Conversion

Example: 10 → 8


```python
oc = 0o10
print(oc)
```

#### Hexadecimal to Decimal Conversion

Example: ff → 255


```python
h = 0xff
print(h)
```

#### Unsing Formate Function

```python
num = 204

print(f'Binary :', format(num, 'b'))
print(f'Octal :', format(num, 'o'))
print(f'Hexadecimal :', format(num, 'x'))
```
