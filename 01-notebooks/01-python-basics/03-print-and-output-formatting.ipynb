## Print And Output Formatting


#### Printing Text

`print("Python")`


#### Printing Multiple Values

`print("Age:", 22)`


#### Using sep in print()

`print(2026, 7, 28, sep="-")` → `2026-7-28`


#### Using end in print()

`print("Hello", end=" ")`


#### Escape Sequence for New Line

`"Hello\nWorld"`


#### Escape Sequence for Tab

`"Name\tAge"`


#### Printing Quotes Inside Strings

`print("He said \"Hello\"")`


#### Formatting with format()

`"My name is {}".format(name)`


#### Formatting with f-Strings

`f"My name is {name}"`
