#### Integer Variables

```python
age = 22
print(age)
```

#### Floating-Point Variables

```python
raduis = 16.56
print(raduis)
```

#### String Variables

```python
name = 'Anshuman'
print(name)
```

#### Boolean Variables

```python
is_active = True
print(is_active)
```

#### Dynamic Typing

```python
value = 10
value = 'Python'

print(value)
```

#### Checking Data Type with type()

```python
print(type(3.14))
print(type(3))
print(type('Anshuman'))
print(type(True))
print(type(34 + 5j))
print(type((1,2,3,4,5)))
print(type( [1,2,3,3,8] ))
print(type( {5,5,2,6,2,8} ))
```
