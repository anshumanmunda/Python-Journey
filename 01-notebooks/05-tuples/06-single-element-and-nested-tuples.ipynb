## Single Element And Nested Tuples


#### Single-Element Tuple Syntax


```python
t = (10,)
print(t, type(t))
```

#### Nested Tuple Creation

Create one tuple creation inside another and display the result.


```python
t = (10, ('Hii', 'How', 'are', 'you'), 90, 30)
print(t)
```

#### Accessing Nested Tuples

Retrieve the required nested tuples and display it.


```python
t = (10, ('Hii', 'How', 'are', 'you'), 90, 30)

print(t[1])
print(t[1][0])
```
