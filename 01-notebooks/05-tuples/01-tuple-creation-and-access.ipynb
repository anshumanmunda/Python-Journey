## Tuple Creation And Access


#### Creating an Empty Tuple

Create an empty tuple and display it.


```python
t1 = tuple()
t2 = ()

print(t1, type(t1))
print(t2, type(t2))
```

#### Creating a Single-Element Tuple

Create a tuple containing only one value.


```python
t = (69,)

print(t, type(t))
```

#### Creating a Multi-Element Tuple

Create multi-element tuple with suitable sample values and display the result.


```python
t = (69, True, 'Hello', 'Lunch', 'Dinner')

print(t)
```

#### Positive Indexing

Display the first and third elements of a list using positive indexes.


```python
t = (69, True, 'Hello', 'Lunch', 'Dinner')

print(t[0])
print(t[2])
```

#### Negative Indexing

Display the last and second-last elements of a list.


```python
t = (69, True, 'Hello', 'Lunch', 'Dinner')

print(t[-1])
print(t[-2])
```
