## Tuple Conversion And User Input


#### List to Tuple Conversion

Convert a suitable value into list to tuple form and display the result.


```python
l = [1,3,4,5,6,True]

t = tuple(l)
t
```

#### String to Tuple Conversion

Convert a suitable value into string to tuple form and display the result.


```python
t = tuple('Python')
t
```

#### Taking Tuple Input


```python
t = tuple( input('Enter elements of tuple seperated by space').split())
print(t, type(t))
```

#### Tuple to List Conversion

Convert a suitable value into tuple to list form and display the result.


```python
t = ('P', 'y', 't', 'h', 'o', 'n')
l = list(t)

l
```
