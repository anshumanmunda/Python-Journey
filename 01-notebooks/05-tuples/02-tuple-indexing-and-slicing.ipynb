## Tuple Indexing And Slicing


#### Tuple Indexing

Indexing is used to access individual elements by position.


```python
t = (69, True, 'Hello', 'Lunch', 'Dinner')

print(t[0])
print(t[4])
```

#### Tuple Slicing

Slicing extracts a selected part of a sequence.


```python
t = (69, True, 'Hello', 'Lunch', 'Dinner')

print(t[0:4])
```

#### Reverse Tuple Slicing

Display tuple slicing in reverse order.


```python
t = (69, True, 'Hello', 'Lunch', 'Dinner')

print(t[::-1])
```

#### Nested Tuple Access

Create one tuple access inside another and display the result.


```python
t = (69, True, 'Hello', 'Lunch', 'Dinner', (1,2,3,4,5,6))

print(t[5])
print(t[5][5])
```
