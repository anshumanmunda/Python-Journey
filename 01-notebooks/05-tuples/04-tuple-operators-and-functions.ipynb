## Tuple Operators And Functions


#### Tuple Concatenation


```python
t1 = (1,2,3,4,5)
t2 = (True, False, 23+6j, None)

print(t1 + t2)
```

#### Tuple Repetition


```python
t1 = (1,2,3,4,5) * 3
t1
```

#### Membership in Tuples


```python
t2 = (10, 20, 30, True, False, 23+6j, None)

print(10 in t2)
```

#### len() with Tuples

`len()` returns the total number of elements in a tuple.


```python
t2 = (10, 20, 30, True, False, 23+6j, None)
len(t2)
```

#### min() with Tuples

`min()` returns the smallest tuple value.


```python
t = (10, 20, 30, 40, 70, -65, 421)
min(t)
```

#### max() with Tuples

`max()` returns the largest tuple value.


```python
max(t)
```

#### sum() with Tuples

`sum()` adds all numeric values in the tuple.


```python
sum(t)
```

#### sorted() with Tuples

`sorted()` returns the tuple values as a new sorted list.


```python
sorted(t)
```
