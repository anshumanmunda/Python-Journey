## Tuple Packing And Unpacking


#### Tuple Packing

Pack three values into one tuple.


```python
t = (10, 20, 30)
print(t, type(t))
```

#### Tuple Unpacking

Unpack three tuple values into separate variables.


```python
x, y, z = (10, 20, 30)

print(x, y, z)
```

#### Extended Tuple Unpacking

Unpacking assigns collection elements to separate variables.


```python
x, *y = (10, 20, 30, 40, 50, 60)
print(x)
print(y)
```

```python
x, y, *z = (10, 20, 30, 40, 50, 60)
print(x)
print(y)
print(z)
```

#### Swapping with Tuple Unpacking

Unpacking assigns collection elements to separate variables.


```python
x = 10 
y = 69

print(f'Before swapping\nx = {x}\ny = {y}')
x, y = 69, 10
print(f'After swapping\nx = {x}\ny = {y}')
```
