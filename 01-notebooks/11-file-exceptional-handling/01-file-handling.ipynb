```python
f = open('sample.txt', 'r')
line1 = f.readline()
line2 = f.readline()
print(line1,line2)
```

readline() using loop

```python
f = open('sample.txt', 'r')

while True:
  data = f.readline()

  if data == "":
    break
  else:
    print(data, end=' ')
```

Chunk-Based Reading Loop

```python
chunk_size = 11

with open('big.txt', 'r') as f:
  data = f.read(chunk_size)

  while  len(data) > 0:
    print(data, end='')
    data = f.read(chunk_size)
```

Serializing a List to JSON File

```python
import json

l = [1,2,3,4,5,6,7,8]
with open('demo.json', 'w') as f :
  json.dump(l,f)
```

Deserializing JSON File Back to Python

```python
with open('demo.json', 'r') as f:
  data = json.load(f)

print(data)  
print(type(data))  
```

 Custom Objects Are Not JSON Serializable by Default

```python
class Person:
  def __init__(self, name: str, age: int, gender: str):
    self.name = name
    self.age = age
    self.gender = gender 

p = Person('Anshuman', 22, 'Male')
```

```python
with open('demo3.json', 'w' ) as f:
  json.dump(p,f)
```

Writing a custom serialization logic

```python
def show_object(person):
  if isinstance(person, Person):
    return {'name': 'Anshuman',
            'age': 22,
            'gender': 'Male'}

with open('demo3.json', 'w') as f:
  json.dump(p, f, default=show_object, indent=4 )  
```

```python
with open('demo3.json', 'r')  as f:
  data = json.load(f)
  print(data)
  print(type(data))
```

Final code of serialization of custom object 

```python
class Person:
  def __init__(self, name: str, age: int, gender: str):
    self.name = name
    self.age = age
    self.gender = gender

  def display_info(self):
    print(f'Name: {self.name}')  
    print(f'Age: {self.age}')  
    print(f'Gender: {self.gender}')  

  def show_object(person):  
    if isinstance(person, Person):
      return {
              'name': 'Anshuman',
              'age': 22,
              'gender': 'Male'
              }

p = Person('Anshuman', 22, 'Male')
with open('demo3.json', 'w') as f:
  json.dump(p, f, default=show_object, indent=4 )

p.display_info()  
```

Pickling a Custom Object

```python
class Person:
  def __init__(self, name: str, age: int, gender: str):
    self.name = name
    self.age = age
    self.gender = gender

  def display_info(self):
    print(f'Name: {self.name}')  
    print(f'Age: {self.age}')  
    print(f'Gender: {self.gender}')  


p = Person('Anshuman', 22, 'Male')
```

```python
import pickle 
with open('objects.pkl', 'wb') as f:
  pickle.dump(p, f)
```

```python
import pickle 
with open('objects.pkl', 'rb') as f:
  p = pickle.load(f)
  p.display_info()

isinstance(p, Person) 
```
