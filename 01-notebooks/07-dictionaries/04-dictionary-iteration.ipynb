## Dictionary Iteration


#### Iterating over Keys

Display every key in the dictionary.


```python
my_dict = {
  'name': 'Anshuman',
  'age': 21,
  'address': {
    'city': 'Ranchi',
    'distic': 'Ranchi',
    'pincode': 834001,
  }
}

for key in my_dict:
  print(key)
```

#### Iterating over Values

Display every value in the dictionary.


```python
for value in my_dict.values():
  print(value)
```

#### Iterating over Items

Display each key together with its value.


```python
for key, value in my_dict.items():
  print(f'{key} -> {value}')
```

#### Iterating with key-value Unpacking

Unpacking assigns collection elements to separate variables.


```python
name, age, address = my_dict.values()
print(name)
print(age)
print(address)
```
