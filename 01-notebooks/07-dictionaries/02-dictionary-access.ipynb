## Dictionary Access


#### Accessing Values by Key

Display the value stored under a selected key.


```python
d = {
    'maths': 90, 
    'physics': 50, 
    'chemistary': 60, 
    'english': 88, 
    'computer': 60
  }

print(d['maths'])
```

#### Accessing with get()

Retrieve the required with get() and display it.


```python
d.get('computer')
```

#### Checking Key Membership

Check key membership and display the result.


```python
'history' in d
```

#### Accessing Nested Dictionary Values

Retrieve the required nested dictionary values and display it.


```python
my_dict = {
  'name': 'Anshuman',
  'age': 22,
  'address': {
    'city': 'Ranchi',
    'distic': 'Ranchi',
    'pincode': 834001,
  }
}

my_dict['address']['city']
```
