## Add Update And Delete Items


#### Adding a New Key

Add a new key-value pair to an existing dictionary.


```python
my_dict = {
  'name': 'Anshuman',
  'age': 21,
  'address': {
    'city': 'Ranchi',
    'distic': 'Ranchi',
    'pincode': 834001,
  }
}

my_dict['course'] = 'MCA'
my_dict
```

#### Updating an Existing Value

Change the value stored under an existing key.


```python
my_dict['age'] = 22
my_dict
```

#### Updating with update()

Change with update() and display the updated result.


```python
my_dict.update(name = 'pankaj')
my_dict
```

#### Deleting with del

Remove with del and display what remains.


```python
my_dict
del my_dict['city']
```

#### Deleting with pop()

Remove with pop() and display what remains.


```python
my_dict

my_dict.pop('address')
```

#### Deleting the Last Item with popitem()

Remove the last item with popitem() and display what remains.


```python
my_dict.popitem()
```

#### Clearing a Dictionary


```python
my_dict

my_dict.clear()
```
