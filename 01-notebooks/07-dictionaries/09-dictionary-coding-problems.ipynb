## Dictionary Coding Problems


#### Character Frequency Dictionary


```python
text = 'I love Python Programming'

my_dict = {}

for ch in text:
  if ch not in my_dict and ch != ' ':
    my_dict[ch] = text.count(ch)

# Printing

for k,v in my_dict.items():
  print(f'{k} -> {v}')

  
```

#### Word Frequency Dictionary


```python
sentence ="Travis Oliphant is the primary developer and creator of NumPy." 

my_dict2 = {}

word_lst = sentence.split(' ')

for word in word_lst:
  if word not in my_dict2 and word != ' ':
    my_dict2[word] = sentence.count(word)

my_dict2    
```

#### Inverting Keys and Values


```python
student_marks = {
    "maths": 90,
    "physics": 50,
    "chemistry": 60
}

inverted = {key : value for value, key in student_marks.items()}
inverted
```

#### Grouping Values by Key


```python
data = [
    ("maths", 90),
    ("physics", 70),
    ("maths", 85),
    ("physics", 75),
    ("computer", 95)
]

grouped = {}

for key, value in data:
  grouped.setdefault(key, []).append(value)

grouped  
```
