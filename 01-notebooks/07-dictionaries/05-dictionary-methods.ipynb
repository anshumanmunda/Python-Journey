## Dictionary Methods


#### keys()

`keys()` returns a dynamic view of dictionary keys.


```python
my_dict = {
  'name': 'Anshuman',
  'age': 21,
  'address': {
    'city': 'Ranchi',
    'distic': 'Ranchi',
    'pincode': 834001,
  }
}

my_dict.keys()
```

#### values()

`values()` returns a dynamic view of dictionary values.


```python
my_dict.values()
```

#### items()

`items()` returns dictionary key-value pairs.


```python
my_dict.items()
```

#### get()

`get()` reads a dictionary value without raising an error for a missing key.


```python
my_dict.get('name')
```

#### setdefault()

`setdefault()` returns a key's value and inserts a default when the key is absent.


```python
my_dict
my_dict.setdefault('city', 'Ranchi')
my_dict
```

#### update()


```python
my_dict.update(city = 'kolkata')
my_dict
```

#### copy()

`copy()` creates a shallow copy of a collection.


```python
copy_dict = my_dict.copy()
copy_dict
```

#### fromkeys()

`fromkeys()` creates a dictionary from a sequence of keys.


```python
my_dict.fromkeys(['english', 'maths', 'phycis'], 0 )
```

```python
subjects = ['english', 'maths', 'phycis']

my_dict.fromkeys(subjects, )
```
