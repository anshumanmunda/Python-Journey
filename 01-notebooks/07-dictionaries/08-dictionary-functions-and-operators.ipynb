## Dictionary Functions And Operators


#### len() with Dictionaries

`len()` returns the total number of key-value pairs in a dictionary.


```python
d = {
  'English' : 78,
  'maths' : 80,
  'computer' : 77,
}

len(d)
```

#### Membership with in


```python
'english' in d
```

#### Dictionary Equality


```python
d = {
  'English' : 78,
  'maths' : 80,
  'computer' : 77,
}
# d3 = {
#   'English' : 78,
#   'maths' : 80,
#   'computer' : 77,
# }

d2 = {
  'name': 'Anshuman',
  'age': 22,
  'city': 'Ranchi',
}

d == d2
```

#### Merging Dictionaries


```python
dict1 = {"name": "Anshuman", "course": "MCA"}
dict2 = {"marks": 90, "semester": 4}

result = dict1 | dict2

print(result)
```
