## Nested Dictionaries


#### Creating a Nested Dictionary

Create nested dictionary with suitable sample values and display the result.


```python
my_dict = {
  'name': 'Anshuman',
  'age': 21,
  'address': {
    'city': 'Ranchi',
    'distic': 'Ranchi',
    'pincode': 834001,
  }
}
```

#### Accessing Nested Values

Retrieve the required nested values and display it.


```python
my_dict.get('address')
```

```python
my_dict.get('address')['pincode']
```

#### Iterating through Nested Dictionaries


```python
for k,v in my_dict.items():
  print(f'{k} -> {v}')
```
