#### Creating from Two Sequences

Construct from two sequences with representative values from the notes.


```python
# Method 1

subject = ['maths', 'physics', 'chemistary', 'english', 'computer']
marks = [90, 50, 60, 88, 60]

new_dict = {}

for i in range( len(subject) ):
  new_dict[subject[i]] = marks[i]

new_dict
```

```python
# Method 2

subject = ['maths', 'physics', 'chemistary', 'english', 'computer']
marks = [90, 50, 60, 88, 60]

new_dict = dict(zip(subject, marks))
new_dict
```
