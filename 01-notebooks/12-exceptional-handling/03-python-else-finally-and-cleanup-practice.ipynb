## Python Else, Finally, and Cleanup Practice


#### Use try-except-else for Division

Perform division inside `try`.

Handle zero division in `except` and display the result only in `else`.


```python
try:
  result = 589 // 10

except ZeroDivisionError:
  print('can not divide by zero')

else:
  print(result)
  print('Divison Sucessfull')
```

#### Use else for Successful Integer Conversion

Convert a suitable value into use else for successful integer form and display the result.


```python
try:
  result = int('10')

except ValueError:
  print('Invlid type conversion')

else:
  print(result)
```

#### Read a File with else

Open a file in `try`, handle a missing file in `except`, and read its content in `else`.


```python
try:
  f = open('sample.txt', 'r') 

except FileNotFoundError:
  print('File not found')

else:
  data = f.read()
  print(data)
```

#### Verify else Does Not Run After an Error

Create a program where the `try` block fails and confirm that `else` is skipped.


```python
try:
  result = 589 // 0

except ZeroDivisionError:
  print('can not divide by zero')

else:
  print(result)
  print('Divison Sucessfull')
```

#### Use finally with Successful Code

Write a successful `try` block and confirm that `finally` still runs.


```python
try:
  f = open('sample.txt', 'r') 

except FileNotFoundError:
  print('File not found')

else:
  data = f.read()
  print(data)

finally:
  f.close()
  print('Program ended sucessfully')  
```
