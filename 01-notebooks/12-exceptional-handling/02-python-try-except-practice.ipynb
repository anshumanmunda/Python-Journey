## Python Try-Except Practice


#### Handle Division by Zero

Use `try-except` to handle division by zero and display a friendly message.


```python
try:
  num = int(input('Enter a number: '))
  print(num // 0)
except ZeroDivisionError:
  print('Can not divide number by zero')  
```

#### Handle Invalid Integer Input

Ask the user for a number and handle `ValueError` if the input cannot be converted to an integer.


```python
try:
  num = int(input('Enter a number: '))
  print(num // 0)
except ZeroDivisionError:
  print('Can not divide number by zero')  
except ValueError:
  print('Please enter a valid number')  
```

#### Handle a Missing File

Try opening a file that does not exist and handle `FileNotFoundError`.


```python
try:
  with open('sample.txt' 'r') as f:
    data = f.read()
    print(data)
except FileNotFoundError:
  print('File not found')
```

#### Create and Read a File Safely

Create a text file, then read it inside a `try-except` block.


```python
# Creating a file

with open('sample.txt', 'w') as f:
  f.write('I love Python Programming.')

# Opening file safely 

try:
  with open('sample.txt','r') as f:
    data = f.read()
    print(data)
except FileNotFoundError:
  print('File not found')
```

#### Catch the Exception Object

Use `except Exception as e` to capture and display the actual error message.


```python
try:
  with open('sample2.txt','r') as f:
    data = f.read()
    print(data)
except Exception as e:
  print(e)
```

#### Handle IndexError Specifically

Access an invalid list index and catch only `IndexError`.


```python
try:
  l = [10, 20, 34, 34, 40]
  l[100]
except IndexError:   
  print('Invalid Index')
```

#### Handle KeyError Specifically

Access a missing dictionary key and catch only `KeyError`.


```python
try:
  my_dict = {
    'name': 'Anshuman',
    'age': 22
  }
  my_dict['city']
except KeyError:   
  print('Invalid Index')
```

#### Handle ValueError Specifically

Convert invalid text to integer and catch only `ValueError`.


```python
try:
  int('a')
except ValueError:
  print('Invalid type conversion')  
```

#### With generic error handler
