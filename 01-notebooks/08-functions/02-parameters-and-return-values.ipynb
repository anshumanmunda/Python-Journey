## Parameters And Return Values


#### Single Parameter

Create a function that receives one number and returns its square.


```python
def greet(name):
  print(f'Welcome {name}')

greet('Anshuman')  
```

#### Multiple Parameters

Create a function that receives two numbers and returns their sum.


```python
def add(a, b , c):
  print(f'sum = {a + b + c}')

add(10,20,30)  
```

#### Returning One Value

Return the result of a calculation from a function.


```python
def add(a, b , c):
  return a + b + c

result = add(10,20,30)
print(result)  
```

#### Returning Multiple Values

Return the minimum and maximum values from a function.


```python
def sum_avg(a, b , c):
  sum = a + b + c
  avg = sum / 3
  return sum, avg

result = sum_avg(10,20,30)
print(f'Sum = {result[0]}') 
print(f'Average = {result[1]}') 
```
