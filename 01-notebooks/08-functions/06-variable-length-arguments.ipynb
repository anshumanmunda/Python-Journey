## Variable Length Arguments


#### Variable Positional Arguments


```python
def average(*args):
  return round(sum(args) / len(args), 2)

result = average(1,2,4,5,6,7,8)
result
```

#### Combining Fixed and Variable Arguments


```python
def fun(name, *args):
  print(name)
  print(args)

fun('Anshuman', 45, 50, 30, 80)
```

#### Variable length  keyword arguments

```python
def marks(**kwargs):
  print(kwargs)

marks(english = 60, maths = 90, physics = 78 )
```
