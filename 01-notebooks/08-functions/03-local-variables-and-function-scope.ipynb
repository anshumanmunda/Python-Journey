## Local Variables And Function Scope


#### Local Variable

This variable stores data at the scope indicated by the heading.


```python
# Global variable
x = 10

def fun():
  # Local variable
  name = 'Python'

  print(name)

fun()  
```

#### Accessing a Global Variable

Retrieve the required a global variable and display it.


```python
print(x)
```

#### Local and Global Names with Same Identifier


```python
# Global variable
x = 10

def fun():
  # Local variable
  x = 'Python'

  print(x)

fun() 
print(x)
```

#### Function Parameter Scope

Scope determines where a name can be accessed in a program.
