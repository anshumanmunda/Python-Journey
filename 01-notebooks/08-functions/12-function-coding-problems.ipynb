#### Function for Prime Check


```python
is_prime = True

def check_prime(n):
  if n < 2:
    print('Not a Prime number')
    return

  for i in range(2,n):
    if n % i == 0:
      print('Not a Prime number')
      break
  else:
      print('Prime number')

check_prime(11)
```

#### Function for Factorial


```python
def factorail(n):
  fact = 1

  for i in range(1, n +1):
    fact *= i

  print(f'{n}! = {fact}')  

factorail(5)
```

#### Function for Fibonacci Sequence


```python
def fibonacci(n):
  first = -1
  second = 1
  for i in range (n):
    next = first + second
    print(next, end=' ')
    first, second = second, next

fibonacci(10)    
```

#### Function for Maximum of Three Values


```python
def maximum(a,b,c):
  t = (a,b,c)
  return max(t)

maximum(10, 20, 90)
```

#### Function for Palindrome Check


```python
def palindrome(num): 
  # Reverse number
  temp = num
  rev = 0  
  while temp :
    rev = 10 * rev  + temp % 10   
    temp //= 10

  if num == rev:
    print('Palidrome')
  else:
    print('Not Palidrome')

palindrome(141)    
```
