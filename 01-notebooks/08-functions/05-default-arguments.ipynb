## Default Arguments


#### Single Default Argument

Create a greeting function with a default name.

Call it with and without supplying a name.


```python
def greet(name, msg = 'Welcome' ):
  print(f'{msg} {name}')

greet('Anshuman')  
greet('Rahul', 'Hi' )  
```

#### Multiple Default Arguments

Use several default arguments values in the same example.


```python
def add(a = 0, b = 0, c = 0):
  return a + b + c
```

#### Overriding Default Values


```python
result = add(10, 60)
result
```
