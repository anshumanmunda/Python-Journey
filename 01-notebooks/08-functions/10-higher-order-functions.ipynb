## Higher Order Functions


#### Function Returning Another Function


```python
def another_function():
  print('This is another function')
  

def fun():
  print('This fun function')
  return another_function 

result = fun()
result()
```

#### Function Accepting Another Function


```python
def greet(name):
  print(f'Hello {name}')

def other_function(greet, name):
  greet(name) 
```

#### Storing Functions in a Collection


```python
l  = [fun, other_function, another_function, greet] 

l[-1]('Anshuman')
```
