## Functions As Arguments


#### Passing a Function as an Argument

Pass a function into another function and use it on a value.


```python
def greet(name):
  print(f'Hello {name}')

def other_function(greet, name):
  greet(name) 
```

#### Calling the Passed Function


```python
other_function(greet, 'Anshuman') 
```

#### Returning a Function

An outer function can return an inner function object.


```python
def another_function():
  print('This is another function')
  

def fun():
  print('This fun function')
  return another_function 

result = fun()
result()
```
