## Positional And Keyword Arguments


#### Positional Arguments

Call a function by passing values in parameter order.


```python
def add (x, y):
  return x + y

result = add(10, 50) 
print(result)
```

#### Keyword Arguments

Call a function by naming each argument explicitly.


```python
def add (x, y):
  return x + y

result = add(y = 90 , x = 67) 
print(result)
```

#### Mixing Positional and Keyword Arguments


```python
def add (x, y, z):
  return x + y + z

result = add(90, 10, z = 88) 
print(result)
```

#### Argument Order Rules


```python
# You can not have positional arguments after a keyword argument
def add (x, y, z):
  return x + y + z

# This will cause syntax error
result = add(90, z = 88, 24) 
print(result)
```

In case of one positional and one keyword arguments you can not set two values
for one parameter this will give `Run time error`

```python
def add (x, y):
  return x + y 

# This will cause syntax error
result = add(88, x = 90 ) 
print(result)
```
