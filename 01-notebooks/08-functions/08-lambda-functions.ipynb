## Lambda Functions


#### Single-Expression Lambda

Create an anonymous function that returns the square of a number.


```python
square = lambda x : x ** 2 
square(10)
```

#### Lambda with Multiple Arguments


```python
add = lambda a, b, c : a + b + c
add(10, 50, 90)
```

#### Lambda with Conditional Expression


```python
fun = lambda x : x ** 2 if x % 2 == 0 else 0

fun(1)
```

#### Passing Lambda to Another Function


```python
add = lambda x : sum(x)

def fun(add, *values):
  result = add(values)
  print(result)

fun(add, 45, 50, 90)  
```
