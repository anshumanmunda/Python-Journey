## Args And Kwargs


#### Collecting Values with *args

Create a function that accepts any number of numeric values and returns their total.


```python
def add(*args):
  return sum(args)

result = add(10, 50, 60, 90, 33)
result
```

#### Collecting Named Values with **kwargs

Create a function that accepts profile details as named values.


```python
def marks(**kwargs):
  print(kwargs)

marks(english = 60, maths = 90, physics = 78, chemistary = 89 )
```

#### Iterating over args

Display each value in args one by one.


```python
def demo(*args):
  for e in args:
    print(e, end=' ')

demo(4, 16, 16, 8, 89, 25)    
```

#### Iterating over kwargs

Display each value in kwargs one by one.


```python
def marks(**kwargs):
  for k,v in kwargs.items():
    print(f'{k} : {v}')

marks(english = 60, maths = 90, physics = 78, chemistary = 89 )
```

#### Using *args and **kwargs Together

Apply *args and **kwargs Together with a small representative example.


```python
def fun(*args, **kwargs):
  for k,v in kwargs.items():
    print(f'{k} : {v}')

  print(f'\nTotal Marks = {sum(args)}')

fun(80, 50, 60, 90, english = 80, physics = 50, maths = 60, computer = 90)
```
