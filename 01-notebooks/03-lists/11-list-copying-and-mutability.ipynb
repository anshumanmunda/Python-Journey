## List Copying And Mutability


#### Reference Assignment


```python
l1 = [10, 20, 30, 40, 50]

new_lst = l1
```

#### Shallow Copy with copy()

```python
l1 = [10, 20, 30, 40, 50]

new_lst = l1.copy()
new_lst
```

#### Shallow Copy with Slicing



```python
l1 = [10, 20, 30, 40, 50]
new_lst = l1[::-1]
```

#### Comparing id() Values

Compare id() values using suitable values.


```python
print(id(l1))
print(id(new_lst))
```
