## List Methods


#### append()

Use suitable values to complete append() and display the result.


```python
l1 = [1,2,3,4,5]
l1.append(69)
l1
```

#### extend()

Use suitable values to complete extend() and display the result.


```python
l1 = [1,2,3,4,5]
l1.extend([77, 88, 'Hello', True])
l1
```

#### insert()

Use suitable values to complete insert() and display the result.


```python
l = [1, 2, 3, 4, 5, 77, 88, 'Hello', True]
l.insert(0, 'hehe')
l
```

#### remove()

Use suitable values to complete remove() and display the result.


```python
l = ['hehe',1, 2, 3, 4, 5, 77, 88, 'Hello', True]
l.remove('hehe')
l
```

#### pop()

Use suitable values to complete pop() and display the result.


```python
l = [1, 2, 3, 4, 5, 77, 88, 'Hello', True]
l.pop()
```

```python
l = [1, 2, 3, 4, 5, 77, 88, 'Hello', True]
l.pop(3)
```

#### clear()

Use suitable values to complete clear() and display the result.


```python
l = [1, 2, 3, 4, 5, 77, 88, 'Hello', True]
l.clear()
l
```

#### index()

Use suitable values to complete index() and display the result.


```python
l = [1, 2, 3, 4, 5, 77, 88, 'Hello', True]
l.index(77)
```

#### count()

Use suitable values to complete count() and display the result.


```python
l = [1, 2, 3, 4, 5, 77, 88, 'Hello', True]

l.count(4)
```

#### sort()

Use suitable values to complete sort() and display the result.


```python
l = [1, 22, 53, 4, 25, 77, 88]

l.sort()
l
```

#### reverse()

Use suitable values to complete reverse() and display the result.


```python
l = [1, 2, 3, 4, 5, 77, 88]

l.reverse()
l
```

#### copy()

Use suitable values to complete copy() and display the result.


```python
l = [1, 2, 3, 4, 5, 77, 88, 'Hello', True]
l2 = l.copy()
l2
```
