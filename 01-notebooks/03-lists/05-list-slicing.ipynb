## List Slicing


#### Basic List Slicing

Extract a middle portion from a list.


```python
l = [1,2,3,4,5,5,6,6,7,7,78,8,8,9,9,9]

mid  = len(l) // 2
l[mid: mid+3]
```

#### Slicing with Start and Stop

Use suitable values to complete slicing with start and stop and display the result.


```python
l = [10, 20, 30, 40, 50, ['Anshuman', 'Akash', 'Itachi']]
l[4:7]
```

#### Slicing with Step

Use suitable values to complete slicing with step and display the result.


```python
l = [10, 20, 30, 40, 50, ['Anshuman', 'Akash', 'Itachi']]
l[1 : 6: 2]
```

#### Reverse List Slicing

Display the list in reverse order using slicing.


```python
l = [10, 20, 30, 40, 50, ['Anshuman', 'Akash', 'Itachi']]
l[::-1]
```

#### Negative Index Slicing

Use suitable values to complete negative index slicing and display the result.


```python
l = [10, 20, 30, 40, 50, ['Anshuman', 'Akash', 'Itachi']]
l[-1: -4: -1]
```
