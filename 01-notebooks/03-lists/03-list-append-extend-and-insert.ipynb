## List Append Extend And Insert


#### Appending One Element

Add one new value to the end of a list.


```python
lst = [10, 20, 30, 40, 50]

lst.append(69)
lst
```

#### Extending with Multiple Elements

Add multiple values from another collection to the list.


```python
lst = [10, 20, 30, 40, 50]

lst.extend(['Anshuman', 'Akash', '69'])
lst
```

#### Inserting at a Specific Position

Insert a value at index `1` and display the updated list.


```python
lst = [10, 20, 30, 40, 50, 'Anshuman', 'Akash', '69']

lst.insert(3, 'Itachi')
lst
```

#### Appending a Nested List

Use suitable values to complete appending a nested list and display the result.


```python
lst = [10, 20, 30, 40, 50]

lst.append(['Anshuman', 'Akash', 'Itachi'])
lst
```
