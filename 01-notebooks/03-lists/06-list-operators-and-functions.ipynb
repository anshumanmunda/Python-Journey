## List Operators And Functions


#### List Concatenation

Use suitable values to complete list concatenation and display the result.


```python
l1 = [1,2,3,4,5]
l2 = [6,7,8,9,10]

l1 + l2
```

#### List Repetition

Use suitable values to complete list repetition and display the result.


```python
l1 = [1,2,3,4,5]  * 2
l1
```

#### Membership in Lists

Use suitable values to complete membership in lists and display the result.


```python
l1 = [1,2,3,4,5]
print(6 in l1)
print(3 in l1)
```
