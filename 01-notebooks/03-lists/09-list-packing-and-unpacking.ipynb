## List Packing And Unpacking


#### List Packing

Use suitable values to complete list packing and display the result.


```python
l = [1,2,3,4,5, 'Hello', True, False]
```

#### List Unpacking

Use suitable values to complete list unpacking and display the result.


```python
x,y,z = [1,2,3]
print(x, y, z)
```

#### Extended Unpacking with *

Use suitable values to complete extended unpacking with * and display the result.


```python
numbers = [10, 20, 30, 40, 50]

first, *rest = numbers
print(first)
print(rest)
```

```python
numbers = [10, 20, 30, 40, 50]

first, middle, *rest = numbers
print(first)
print(middle)
print(rest)
```

#### Swapping Values through Unpacking

Use suitable values to complete swapping values through unpacking and display the result.


```python
a,b = 10, 20
print(f'Before Swapping\na = {a} b = {b}')

a,b = 20, 10
print(f'After Swapping\na = {a} b = {b}')
```
