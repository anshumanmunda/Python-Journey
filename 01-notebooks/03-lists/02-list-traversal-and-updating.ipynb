## List Traversal And Updating


#### Traversing a List with for

Visit every element in a list with for in sequence.


```python
lst = [10, 20, 30, 40, 50]

for i in lst:
  print(i ,end=' ')
```

#### Traversing a List with while

Visit every element in a list with while in sequence.


```python
lst = [10, 20, 30, 40, 50]

i = 0
while i < len(lst):
  print(lst[i], end=' ') 
  i += 1
```

#### Updating a Single Element

Change a single element and display the updated result.


```python
lst = [10, 20, 30, 40, 50]

lst[3] = 69
lst
```

#### Updating Multiple Elements with Slicing

Change multiple elements with slicing and display the updated result.


```python
lst = [10, 20, 30, 40, 50]

lst[0:3] = ['Hehe', 'haha', 'hehe']
lst
```

#### Updating Nested List Elements

Change nested list elements and display the updated result.


```python
lst = [10, 20, 30, [12, 'Rajesh', 456, 'Akash']]

lst[3][3] = 'Anshuman'
print(lst)
```
