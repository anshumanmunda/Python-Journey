## List Comprehension


#### Basic List Comprehension

Create a list containing the squares of numbers from `1` to `5`.


```python
lst = [i ** 2 for i in range(1,6)]
lst
```

#### List Comprehension with Condition

Create a list containing only even numbers from `1` to `10`.


```python
lst = [i  for i in range(1,11) if i % 2 == 0]
lst
```

#### List Comprehension with if-else

Use suitable values to complete list comprehension with if-else and display the result.


```python
lst = [ i**2 if i % 2 == 0 else i ** 3  for i in range(1,11)  ]
lst
```

```python
lst = ['even' if i % 2 == 0 else 'odd' for i in range(1,6)]
lst
```

#### Nested List Comprehension

Create one list comprehension inside another and display the result.


```python
lst = [[ i * j for j in range(1,4)]for i in range(1,4) ]
lst
```

#### Flattening a Nested List

Use suitable values to complete flattening a nested list and display the result.


```python
lst = [[1, 2, 3], [2, 4, 6], [3, 6, 9]]

new_list = []
for sub  in lst:
  for e in sub:
    new_list.append(e)

new_list
```

```python
lst = [[1, 2, 3], [2, 4, 6], [3, 6, 9]]

flat_list = [item for sublist in lst for item in sublist]
flat_list
```
