## List Creation And Access


#### Creating an Empty List

Create an empty list and display it.


```python
l1 = list()
l2 = [] 

print(l1)
print(l2)
```

#### Creating a Homogeneous List

Create a list containing only integer values.


```python
lst = [10, 20, 30, 40, 50]
print(lst)
```

#### Creating a Heterogeneous List

Create a list containing an integer, a string, and a Boolean value.


```python
lst = ['Ram', 'Mango', 89, True, 34 + 5j, 23.50]
print(lst)
```

#### Positive Indexing

Display the first and third elements of a list using positive indexes.


```python
l = ['Ram', 'Mango', 89, True, 34 + 5j, 23.50]

print(l[0])
print(l[3])
```

#### Negative Indexing

Display the last and second-last elements of a list.


```python
l = ['Ram', 'Mango', 89, True, 34 + 5j, 23.50]

print(l[-1])
print(l[-2])
```

#### Accessing Nested List Elements

Retrieve the required nested list elements and display it.


```python
l3 = [
  [1,2,34,4],
  [56,45,1,5],
  [86,21,0,9],
  ]

print(l3[1][0])
print(l3[2][2])
print(l3[2][3])
```
