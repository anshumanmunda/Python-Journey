## List Delete And Remove


#### Deleting with del

Remove with del and display what remains.


```python
l = [10, 20, 30, 40, 50, ['Anshuman', 'Akash', 'Itachi']]

del l[5]
l
```

#### Deleting a Slice

Remove a slice and display what remains.


```python
l = [10, 20, 30, 40, 50, ['Anshuman', 'Akash', 'Itachi']]

del l[4:]
l
```

#### Removing by Value with remove()

Remove a selected value from the list.


```python
l = [10, 20, 30, 40, 50, ['Anshuman', 'Akash', 'Itachi']]

l.remove(50)
l
```

#### Removing by Index with pop()

Remove the element at a selected index and display the removed value.


```python
l = [10, 20, 30, 40, 50, ['Anshuman', 'Akash', 'Itachi']]
l.pop()
```

```python
l.pop()
```

```python
l.pop(0)
l
```

#### Removing the Last Element

Use suitable values to complete removing the last element and display the result.


```python
l = [10, 20, 30, 40, 50]

l.pop()
```

#### Clearing a List with clear()

Remove all elements from the list.


```python
l = [10, 20, 30, 40, 50, ['Anshuman', 'Akash', 'Itachi']]

l.clear()
l
```
