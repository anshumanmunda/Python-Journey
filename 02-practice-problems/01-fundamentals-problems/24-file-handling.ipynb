#### 1. Reading a Text File: Write a program that opens a file named data.txt in read mode (r) and prints its entire contents to the console.


```python
with open('assets/data.txt', 'r') as f:
  data = f.read()

print(data)
```

#### 2. Line-by-Line Iteration: Open a file and use a for loop to read it line-by-line. Print each line after stripping the trailing newline character.


```python
with open('assets/data.txt', 'r') as f:
  for line in f:
    print(line, end='')
```

#### 4. Appending Data: Open an existing file in append mode (a) and add a new timestamped log entry to the end of the file without overwriting the original content.

#### 6. Reading as a List: Use the .readlines() method to read a file and store every line as an individual element in a Python list.


```python
lst = []
with open('assets/data.txt', 'r') as f:
  lines = f.readlines()
  for line in lines:
    lst.append(line)

lst
```

#### 7. Checking File Existence: Import the os module and write a function that checks if a specific filename exists on the disk before attempting to open it.


```python
# write your code here
import os

def check_file(file:str)-> None:
  path = 'assets/' + file 
  if os.path.exists(path):
    return path
  else:
    return False

file = check_file('notes.txt')
if file:
  with open(file, 'r') as f:
    data = f.read()
    print(data)
else:
  print(f'File not found')    
```

#### 8. Handling CSV Data: Use file handling (without the CSV module) to read a comma-separated line (e.g., 'John,25,NY') and split it into a list of individual strings.

#### 9. Writing Lists to Files: Given a list of names, use the .writelines() method to save them all into a text file at once.


```python
names = ['Anshuman\n', 'Manish\n', 'Akash\n', 'Ajay\n', 'Vijay\n', 'Shubham\n', 'Rajesh']
with open('assets/names.txt', 'w') as f:
  f.writelines(names)
```

#### 10. Binary File Copying: Demonstrate how to read an image or a PDF file in binary mode (rb) and write its contents to a new file (wb) to create a copy.


```python
with open('assets/anshu_pic.png', 'rb') as f:
  binary_data = f.read()


with open('assets/copy-anshu-pic.png', 'wb') as f:
  f.write(binary_data)   
```
