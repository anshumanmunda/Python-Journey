### `Q1:` Join Tuples if similar initial element
While working with Python tuples, we can have a problem in which we need to perform concatenation of records from the similarity of initial element. This problem can have applications in data domains such as Data Science.

For eg.
```
Input  : test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
Output : [(5, 6, 7, 8), (6, 10), (7, 13)]
```

````python
# write your code here
test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
first_element_lst = []

result_list = []


for item in test_list:
  t = tuple()
  first_element = item[0]

  if first_element not in first_element_lst:
    for i in test_list:
      if i[0] == first_element :
        if i[0] not in t:
          t += (i[0], )
        if i[1] not in t:
          t += (i[1], )

    first_element_lst.append(first_element)
    result_list.append(t)    

result_list
````

### `Q2:` Multiply Adjacent elements (both side) and take sum of right and left side multiplication result.


For eg.
```
The original tuple : (1, 5, 7, 8, 10)
Resultant tuple after multiplication :

(1*5, 1*5+5*7, 7*5 + 7*8, 8*7 + 8*10, 10*8) -> (5, 40, 91, 136, 80)

output-(5, 40, 91, 136, 80)
```

````python
original_tuple = (1, 5, 7, 8, 10)
final_tuple = ()

for index, value in enumerate(original_tuple):
  if index == 0:
    final_tuple += (value * original_tuple[index + 1], )
  elif index == len(original_tuple) - 1 :
    final_tuple += (value * original_tuple[index - 1], )
  else:
    final_tuple += ( value * original_tuple[index - 1] +  value * original_tuple[index + 1],  )

print(final_tuple)
````

### `Q4`: Count no of tuples, list and set from a list
```
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

```
`Output:`

```
List-2
Set-2
Tuples-1
```

````python
# write your code here
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

total_list = total_set = total_tuple = 0

for item in list1:
  if type(item) == list:
    total_list += 1
  elif type(item) == set:  
    total_set += 1
  else:  
    total_tuple += 1

print(f'List = {total_list}')   
print(f'Set = {total_set}')   
print(f'Tupe = {total_tuple}')   
````

```
Input : ar1 = [1, 5, 10, 20, 40, 80]
        ar2 = [6, 7, 20, 80, 100]
        ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

Output : [80, 20]
```

````python
# write your code here
l1 = [1, 5, 10, 20, 40, 80]
l2 = [6, 7, 20, 80, 100]
l3 = [3, 4, 15, 20, 30, 70, 80, 120]
lst = []

for e in l1:
  if e in l2 and e in l3 :
    lst.append(e)
print(sorted(lst, reverse=True))    
````

### `Q3`: Convert List to List of dictionaries. Given list values and keys list, convert these values to key value pairs in form of list of dictionaries.

**Example 1:**

Input:
```bash
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'DataScience', 'id': 3}, {'name': 'is', 'id': 8}]
```

**Example 2:**

Input:
```bash
test_list = ["CampusX", 10]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'CampusX', 'id': 10}]
```

````python
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]

result_list = list()

if len(key_list) < len(test_list):
  key_list = key_list *  (len(test_list) - len(key_list))

for indx, key in enumerate(key_list):
   temp_dict = {}
   temp_dict[key] = test_list[indx]  
   result_list.append(temp_dict)

result_list   
````

Output:


````python
d = {'c': [3], 'b': [12, 10], 'a': [19, 4]}

new_dict = {}

for key in sorted(d):
  new_dict[key] = sorted(d[key])

new_dict
````
