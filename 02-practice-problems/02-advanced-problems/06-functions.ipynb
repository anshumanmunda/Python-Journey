### **`Problem-1:`** Write a Python function that takes a list and returns a new list with unique elements of the first list.

**Exercise 1:**

Input:

```bash
[1,2,3,3,3,3,4,5]
```

Output:

```bash
[1, 2, 3, 4, 5]
```

````python
def unique_list(test_lst : list ) -> list:
  return list(set(test_lst))

result = unique_list([1,2,3,3,3,3,4,5])
result
````

### **`Problem-2:`** Write a Python function that accepts a hyphen-separated sequence of words as parameter and returns the words in a hyphen-separated sequence after sorting them alphabetically.

**Example 1:**

Input:
```bash
green-red-yellow-black-white
```

Output:
```bash
black-green-red-white-yellow
```

````python
def sort_string(s : str) -> str:
  sorted_list = sorted(s.split('-'))
  return '-'.join(sorted_list)

str1 = "green-red-yellow-black-white"
sort_string(str1)
````

### **`Problem 3:`** Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.

```
Sample String : 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
Expected Output :
No. of Upper case characters :  9
No. of Lower case Characters :  47
```

````python
def count_characters(s : str) -> tuple:
  upper_case = lower_case = 0
  for ch in s:
    if ch.isupper():
      upper_case += 1
    elif ch.islower():
      lower_case += 1

  return upper_case, lower_case

str1 = "CampusX is an Online Mentorship Program fOr EnginEering studentS."
upper_case, lower_case = count_characters(str1)
print(f'Upper case characters : {upper_case}')
print(f'Lower case characters : {lower_case}')
````

### **`Problem 5:`** Write a Python function to check whether a number is perfect or not.

A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

Example :

```
The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.

The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.
```


````python
def perfect_number(num : int) -> None:
  sum = 0
  for i in range(1, num + 1):
    if num % i == 0:
      sum += i

  if sum // 2 == num:
    print(f'{num} is a Perfect Number ') 
  else:     
    print(f'{num} is not a Perfect Number ') 

perfect_number(8128)
````

### **`Problem-6:`** Write a Python function to concatenate any no of dictionaries to create a new one.

```
Sample Dictionary :
dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}
Expected Result : {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}
```

````python
def conc_dict(*args: dict) -> dict:
  result_dict = {}

  for new_dict in args:
    for key, value in new_dict.items():
      result_dict[key] = value

  return result_dict


dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}

result = conc_dict(dic1, dic2, dic3)
result
````


`Problem-7` Write a python function that accepts a string as input and returns the word with most occurence.

```
Input:
hello how are you i am fine thank you
```

```
Output
you -> 2
```

````python
def word_occurence(sen: str) -> str:
  ferquency = {}
  for word in sen.split(' '):
    ferquency[word] = ferquency.get(word, 0) + 1

  max_count_key = max(ferquency, key=ferquency.get)
  return max_count_key, ferquency[max_count_key] 

key, value =  word_occurence("hello how are you i am fine thank you")
print(f'{key} -> {value}')
````

`Problem-8` Write a python function that receives a list of integers and prints out a histogram of bin size 10

```
Input:
[13,42,15,37,22,39,41,50]
```

```
Output:
{11-20:2,21-30:1,31-40:2,41-50:3}
```

````python
def histogram(nums):
    bins = {}

    for num in nums:
        start = (num // 10) * 10
        bins[start] = bins.get(start, 0) + 1

    for start in sorted(bins):
        print(f"{start}-{start + 9}: {'*' * bins[start]}")

l = [13,42,15,37,22,39,41,50]
histogram(l)
````

`Problem-9` Write a python function that accepts a list of 2D co-ordinates and a query point, and then finds the the co-ordinate which is closest in terms of distance from the query point.

```
List of Coordinates
[(1,1),(2,2),(3,3),(4,4)]
Query Point
(0,0)
```

```
Output
Nearest to (0,0) is (1,1)
```

````python
def coordinates(lst: list, query_point: tuple) -> None :
  new_distance = []
  for point in lst:
    x =  abs(point[0] - query_point[0])
    y =  abs(point[1] - query_point[1])

    distance = (x ** 2 + y ** 2) ** 0.5 
    new_distance.append( (new_distance, point) )

  result = sorted(new_distance)
  return  result[0][1]
  

lst = [(1,1),(2,2),(3,3),(4,4)]
query_point = (0,0)

result = coordinates(lst, query_point)
print(result)
````

`Problem 15`- A dictionary contains following information about 5 employees:
- First name
- Last name
- Age
- Grade(Skilled,Semi-skilled,Highly skilled)<br>
Write a program using map/filter/reduce to a list of employees(first name + last name) who are highly skilled
